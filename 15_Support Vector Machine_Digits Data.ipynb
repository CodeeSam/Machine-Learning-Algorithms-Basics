# The Use of Decision Classifier Using Sklearn Digits Data

### Importing Pandas Library and Loading Digits Data

```python
import pandas as pd
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
dir(digits)
```

```python
digits.data
```

```python
digits.target_names
```

### Creating Digit DataFrame

```python
df = pd.DataFrame(digits.data, columns=digits.feature_names)
df.head()
```

### Adding Target Column To DataFrame

```python
df['target'] = digits.target
df.head()
```

### Importing Train Test Split from Sklearn

```python
from sklearn.model_selection import train_test_split
```

### Splitting Data Into Training and Testing Data In Ratio 80:20

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop('target', axis=1), df.target, test_size=0.2 )
```

### Length of Training and Testing Data

```python
len(X_train)
```

```python
len(y_train)
```

```python
len(X_test)
```

```python
len(y_test)
```

### Importing Support Vector Machine

```python
from sklearn.svm import SVC
```

```python
model = SVC()
```

### Model Training

```python
model.fit(X_train, y_train)
```

### Model Accuracy

```python
model.score(X_test, y_test)
```

### Model Testing

```python
model.predict(X_test)
```
