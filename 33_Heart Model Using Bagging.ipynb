#### I will create dictionaries for them except for Sex and ExerciseAngina which I'm gonna get dummies for

```python
df4 = df3.copy()
```

```python
df4.ChestPainType.replace({'ATA':1, 'NAP':2, 'ASY':3, 'TA':4}, inplace=True)

df4.RestingECG.replace({'Normal':1, 'ST':2, 'LVH':3}, inplace=True)

df4.ST_Slope.replace({'Up':1, 'Flat':2, 'Down':3}, inplace=True)
```

```python
df4.head()
```

```python
df5 = pd.get_dummies(df4, drop_first=True)
df5.head()
```

```python
X = df5.drop('HeartDisease', axis=1)
X.head()
```

```python
y = df5.HeartDisease
y.head()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scale = MinMaxScaler()
```

```python
X_scaled = scale.fit_transform(X)
X_scaled
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, stratify=y, random_state=10)
```

### Using SVC alone

```python
from sklearn.svm import SVC
```

```python
svm = SVC()
```

```python
svm.fit(X_train, y_train)
```

```python
svm.score(X_test, y_test)
```

```python
from sklearn.model_selection import cross_val_score

score = cross_val_score(SVC(), X, y, cv=5)
score.mean()
```

### Using SVC with Bagging

```python
from sklearn.ensemble import BaggingClassifier
```

```python
BC = BaggingClassifier(
    base_estimator=SVC(),
    n_estimators=100,
    max_samples=0.8,
    oob_score=True,
    random_state=0
)
```

```python
BC.fit(X_train, y_train)
```

```python
BC.score(X_test, y_test)
```

```python
score = cross_val_score(BC, X,y, cv=5)
```

```python
score.mean()
```

### Using  Decision Tree alone

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
DTC = DecisionTreeClassifier()
```

```python
DTC.fit(X_train, y_train)
```

```python
DTC.score(X_test, y_test)
```

```python
score = cross_val_score(DecisionTreeClassifier(), X,y, cv=5)
score.mean()
```

### Using Decision Tree with Bagging

```python
from sklearn.ensemble import BaggingClassifier
```

```python
BC_DTC = BaggingClassifier(
    base_estimator=DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.8,
    oob_score=True,
    random_state=0
)
```

```python
BC_DTC.fit(X_train, y_train)
```

```python
BC_DTC.score(X_test, y_test)
```

```python
score = cross_val_score(BC_DTC, X,y, cv=5)
```

```python
score.mean()
```
