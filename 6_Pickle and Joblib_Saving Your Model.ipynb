```python
import numpy as np
import pandas as pd
from sklearn import linear_model
```

```python
df = pd.read_csv('homeprices.csv')
df
```

```python
model = linear_model.LinearRegression()
model.fit(df[['area']], df.price)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
model.predict([[5000]])
```

```python
import pickle
```

```python
with open('model_pickle', 'wb') as f:
    pickle.dump(model, f)
```

```python
with open('model_pickle', 'rb') as f:
    mp= pickle.load(f)
```

```python
mp.predict([[5000]])
```

```python
mp.coef_
```

```python
mp.intercept_
```
