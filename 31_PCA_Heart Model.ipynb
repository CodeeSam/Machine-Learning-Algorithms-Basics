```python
df2.head()
```

```python
df3 = pd.get_dummies(df2, columns=['Sex', 'ChestPainType', 'RestingECG', 'ExerciseAngina', 'ST_Slope'])
df3
```

```python
df3.isna().any()
#no missing value
```

```python
X = df3.drop('HeartDisease', axis=1)
X
```

```python
y = df3.HeartDisease
y
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scale = MinMaxScaler()
```

```python
X_scaled = scale.fit_transform(X)
X_scaled
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=30)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
RFC = RandomForestClassifier()
```

```python
RFC.fit(X_train, y_train)
RFC.score(X_test, y_test)
```

```python
from sklearn.decomposition import PCA
```

```python
## using pca, give us 95% most useful features and best dimension
pca = PCA(0.95)
X_pca = pca.fit_transform(X)
X_pca
```

```python
X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=30)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
RFC = RandomForestClassifier()
```

```python
RFC.fit(X_train_pca, y_train)
RFC.score(X_test_pca, y_test)
```
