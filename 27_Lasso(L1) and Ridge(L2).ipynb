```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# to ignore the warning so as to have a clean notebook
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('Melbourne_housing_FULL.csv')
df.head()
```

```python
df.shape
```

```python
#df.nunique()
```

```python
cols_to_be_used = ['Suburb', 'Rooms', 'Type', 'Method', 'SellerG', 'Regionname', 'Propertycount', 'Distance', 'CouncilArea',
                  'Bedroom2', 'Bathroom', 'Car', 'Landsize', 'BuildingArea', 'Price']

df = df[cols_to_be_used]
df.head()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
cols_to_fill_zero = ['Propertycount', 'Distance', 'Bedroom2', 'Bathroom', 'Car']

df[cols_to_fill_zero] = df[cols_to_fill_zero].fillna(0)
```

```python
df.isna().sum()
#we no longer have NA values for these above as we have now fill their Na values with zeros
```

```python
# we will fill landsize and building area with their mean
df['Landsize'] = df['Landsize'].fillna(df.Landsize.mean())
df['BuildingArea'] = df['BuildingArea'].fillna(df.BuildingArea.mean())
```

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
# inplace-true, so that the arrangement wont be scattered after removing the na row
```

```python
df.isna().sum()
# we no longer have any na value
# price was also drop since it the dependent variable
```

```python
df.head()
```

```python
## let's do OHE for the needed columns
df = pd.get_dummies(df, drop_first=True)
df.head()
```

```python
X = df.drop('Price', axis=1)
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
regLR = LinearRegression().fit(X_train, y_train)
```

```python
regLR.score(X_test, y_test)
```

```python
# the above score is very low beacuse the dataset was overfitted: LinearRegression().fit(X_train, y_train)
# let's see the score for the training data
```

```python
regLR.score(X_train, y_train)
```

```python
# in order to address this, we use Lasso regression/ L1 regularization
from sklearn.linear_model import Lasso
```

```python
lasso_reg = Lasso(alpha=50, max_iter=100, tol=0.1)
lasso_reg.fit(X_train, y_train)
```

```python
# now let's check the score as we have now used Lasso
lasso_reg.score(X_test, y_test)
```

```python
# the score has now improved compare to 0.13...... we had earlier
```

```python
from sklearn.linear_model import Ridge
#the Ridge above is L2 Regularization; let's try that
```

```python
ridge_reg = Ridge(alpha=50, max_iter=100, tol=0.1)
ridge_reg.fit(X_train, y_train)
```

```python
ridge_reg.score(X_test, y_test)
```

```python
# we also have the same score
```
