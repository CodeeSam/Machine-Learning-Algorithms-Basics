# The Use of K-Fold Cross Validation Using Sklearn Iris Data

### Importing Libraries

```python
import numpy as np
import pandas as pd
```

### Importing Needed Algorithms

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
import warnings
warnings.filterwarnings('ignore')
```

### Loading Iris Data

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
dir(iris)
```

### Importing K Fold Cross Validation

```python
from sklearn.model_selection import cross_val_score
```

```python
LR_scores = cross_val_score(LogisticRegression(), iris.data, iris.target)
LR_scores
```

```python
SVC_scores = cross_val_score(SVC(), iris.data, iris.target)
SVC_scores
```

```python
DTC_scores = cross_val_score(DecisionTreeClassifier(), iris.data, iris.target)
DTC_scores
```

```python
RFC_scores = cross_val_score(RandomForestClassifier(), iris.data, iris.target)
RFC_scores
```

### Calculating the average of each Algorithm

```python
np.average(LR_scores)
```

```python
np.average(SVC_scores)
```

```python
np.average(DTC_scores)
```

```python
np.average(RFC_scores)
```

## Logistic Regression Has The Best Score!
