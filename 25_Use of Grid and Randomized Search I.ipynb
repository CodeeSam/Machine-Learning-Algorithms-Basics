```python
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
dir(iris)
```

```python
df = pd.DataFrame(iris.data, columns= iris.feature_names)
df
```

```python
df['target'] = iris.target
df['target'] = df['target'].apply(lambda x: iris.target_names[x])
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('target', axis=1)
X.head()
```

```python
y = df.target
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
cross_val_score(SVC(kernel='rbf', C=10, gamma='auto'), iris.data, iris.target, cv=5)
```

```python
# we can also run the above using different parameters at once using loop rather than changing the parameter one by one
```

```python
import numpy as np
kernel = ['rbf', 'linear']
C = [1, 10, 20]
avg_scores = {}

for kv in kernel:
    for cv in C:
        cv_scores = cross_val_score(SVC(kernel=kv, C=cv, gamma='auto'), iris.data, iris.target, cv=5)
        avg_scores[kv + '_' + str(cv)] = np.average(cv_scores)
avg_scores
```

```python
## the above stuff can be easily done without using loop with the aid of Grid Search CV
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
clf = GridSearchCV(SVC(gamma = 'auto'), {
    'kernel' : ['rbf', 'linear'],
    'C' : [1, 10, 20]
},
cv=5, return_train_score=False)
```

```python
#the next thing is to fit your dataset... unlike crossvalscore where you don't need to fit again after supplying the X and y params inside your parenthesis
clf.fit(X,y)
```

```python
#after fitting, check out your results
#clf.cv_results_
```

```python
df = pd.DataFrame(clf.cv_results_)
df.head()
```

```python
df = df[['param_C', 'param_kernel', 'mean_test_score']]
df
```

```python
model_params = {
    'svm':{
        'model':SVC(gamma = 'auto'),
        'params': { 'C':[1,10,20], 'kernel':['rbf', 'linear']}
    },
    'random_forest':{
        'model': RandomForestClassifier(),
        'params': {'n_estimators':[1,5,10]}
    },
    'logistic_regression':{
        'model': LogisticRegression(solver='liblinear', multi_class='auto'),
        'params': {'C': [1,5,10]}
    }
}
```

```python
scores = []

for mn, mp in model_params.items():
    RSCV = GridSearchCV(mp['model'], mp['params'], cv=5, return_train_score=False)
    RSCV.fit(iris.data, iris.target)
    scores.append({
    'model': mn,
    'best_score': RSCV.best_score_,
    'best_params': RSCV.best_params_
    })
```

```python
df = pd.DataFrame(scores, columns=['model', 'best_score', 'best_params'])
df
```
