### Importing Libraries

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
```

### Loading csv file

```python
df = pd.read_csv('homeprices.csv')
df
```

## Data Cleaning
### Handling missing value(s)

```python
import math
bedroom_median = df.bedrooms.median()
bedroom_median
```

```python
df.bedrooms.fillna(bedroom_median)
```

```python
##saving changes to file
```

```python
df.bedrooms = df.bedrooms.fillna(bedroom_median)
df
```

### Model Training

```python
reg = linear_model.LinearRegression()
```

```python
reg.fit(df[['area','bedrooms','age']], df.price)
## the independent variables are fitted in the square brackets
```

### Model Testing

```python
reg.coef_
```

```python
reg.intercept_
```

```python
reg.predict([[3000,3,40]])
```
