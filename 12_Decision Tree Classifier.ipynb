```python
import pandas as pd
```

```python
df = pd.read_csv('salaries.csv')
df.head()
```

```python
# X(independent variable)
inputs = df.drop('salary_more_then_100k', axis=1)
inputs
```

```python
# y(dependent variable)
target = df['salary_more_then_100k']
target.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
ohe = LabelEncoder()
```

```python
inputs['company_n'] = ohe.fit_transform(inputs['company'])
inputs['job_n'] = ohe.fit_transform(inputs['job'])
inputs['degree_n'] = ohe.fit_transform(inputs['degree'])
```

```python
inputs.head()
```

```python
inputs_n = inputs.drop(['company', 'job', 'degree'], axis=1)
inputs_n
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(inputs_n, target)
```

```python
a = df.drop(['company', 'job', 'degree'], axis=1)
a.head()
```

```python
frames = [inputs_n, a]
b = pd.concat(frames, axis=1)
b
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(b[['company_n', 'job_n', 'degree_n']], b.salary_more_then_100k, test_size=0.1)
```

```python
#we have alraedy done the training using decision tree classifier, so let's predict now
model.predict(X_test)
```
