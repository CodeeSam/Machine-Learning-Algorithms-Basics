# The Use Of Random Forest Classifier: Sklearn Iris Data

### Importing Pandas Library and Loading Iris Data

```python
import pandas as pd
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
dir(iris)
```

### Creating Iris DataFrame

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df
```

### Adding Target Column To DataFrame¶

```python
df['target'] = iris.target
df
```

```python
iris.target_names
```

### Importing Train Test Split from Sklearn

```python
from sklearn.model_selection import train_test_split
```

### Splitting Data Into Training and Testing Data In Ratio 90:10

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop('target', axis=1), df.target, test_size=0.1)
```

### Importing Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier()
```

### Model Training

```python
model.fit(X_train, y_train)
```

### Model Accuracy

```python
model.score(X_test, y_test)
```

### Model Testing

```python
y_predicted = model.predict(X_test)
y_predicted
```

##### Actual Values of y

```python
y_actual = y_test
y_actual
```

### Using Confusion Matrix To Check Correlation Between predicted and Actual Values

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_actual, y_predicted)
cm
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(cm, annot=True)
plt.xlabel('Predicted')
plt.ylabel('Actual')
```
